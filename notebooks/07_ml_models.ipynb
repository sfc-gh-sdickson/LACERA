## Model 1: Return Forecasting

**Methodology:** Exponential smoothing with momentum adjustment

**Features:**
- 12-month trailing average returns
- Momentum factor (rate of change in average)
- Historical volatility for confidence intervals

**Output:** Forecasted monthly returns with 95% confidence bands

**Function:** `LACERA_DB.ANALYTICS.FORECAST_RETURNS(asset_class, horizon_months)`

```python
-- Example usage of FORECAST_RETURNS
-- SELECT FORECAST_RETURNS('US Equity', 6);
```

## Model 2: Risk Anomaly Detection

**Methodology:** Statistical Z-score outlier detection

**Features:**
- Rolling 12-month mean and standard deviation
- Z-score calculation for volatility, tracking error, and max drawdown
- Threshold: |Z| > 2.0 for flagging anomalies

**Severity Levels:**
- CRITICAL: |Z| > 3.0
- HIGH: |Z| > 2.5
- MEDIUM: |Z| > 2.0

**Function:** `LACERA_DB.ANALYTICS.DETECT_RISK_ANOMALIES(lookback_days)`

```python
-- Example usage of DETECT_RISK_ANOMALIES
-- SELECT DETECT_RISK_ANOMALIES(90);
```

## Model 3: Manager Performance Scoring

**Methodology:** Multi-factor weighted composite scoring

**Scoring Factors:**
- Return Performance (30%): Percentile rank of average 1Y return
- Risk-Adjusted Return (30%): Percentile rank of Sharpe ratio
- Consistency (25%): Inverse percentile of return volatility
- Fee Efficiency (15%): Excess return per basis point of fee

**Output:** Ranked list of managers with component scores and quartile assignment

**Function:** `LACERA_DB.ANALYTICS.SCORE_MANAGER_PERFORMANCE()`
